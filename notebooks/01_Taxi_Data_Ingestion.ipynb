```python
from pyspark.sql import functions as F 
file_path = "/Volumes/Learn_get_started_de_palvikrant235/taxi_etl_project/raw_files/NYC.csv"
df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(file_path)
)
display(df.count()) 
```

# Null check

```python
df.select([
    F.count(
        F.when(F.col(c).isNull(), c)
    ).alias(c)
    for c in df.columns
]).display()
```

```python
from pyspark.sql import functions as F

bronze_table = "learn_get_started_de_palvikrant235.taxi_etl_project.bronze_taxi_trips"

df_bronze = spark.table(bronze_table)

print("Total Records:", df_bronze.count())
print("Columns:", df_bronze.columns)

df_bronze.printSchema()
```

```python
df_bronze_v2 = (
    df_bronze
    .withColumn(
        "batch_id",
        F.lit("INITIAL_LOAD_001")
    )
    .withColumn(
        "source_file",
        F.lit("train.csv")
    )
    .withColumn(
        "source_system",
        F.lit("KAGGLE")
    )
    .withColumn(
        "ingestion_timestamp",
        F.current_timestamp()
    )
)
```
