```python
from pyspark.sql import functions as F

source_table = "learn_get_started_de_palvikrant235.taxi_etl_project.bronze_taxi_trips"

df_source = spark.table(source_table)

print("Source Records:", df_source.count())
```

```python
df_batched = (
    df_source
    .withColumn(
        "batch_number",
        F.pmod(F.xxhash64("id"), F.lit(3))
    )
)

display(df_batched.limit(10))
```

```python
df_test_batch = (
    df_batched
    .filter(F.col("batch_number") == 0)
    .limit(1000)
    .drop("batch_number")
    .withColumn("batch_id", F.lit("BATCH_004"))
    .withColumn("source_file", F.lit("NYC.csv"))
    .withColumn("source_system", F.lit("KAGGLE"))
    .withColumn("ingestion_timestamp", F.current_timestamp())
)

print("New Test Batch Records:", df_test_batch.count())
```

```python
silver_table = "learn_get_started_de_palvikrant235.taxi_etl_project.silver_taxi_incremental"

df_existing_ids = spark.table(silver_table).select("id").distinct()

df_new_ids = (
    df_test_batch
    .select("id")
    .join(df_existing_ids, on="id", how="left_anti")
)

print("Total Test Records:", df_test_batch.count())
print("New IDs:", df_new_ids.count())
```

```python
from delta.tables import DeltaTable

delta_target = DeltaTable.forName(
    spark,
    silver_table
)

(
    delta_target.alias("target")
    .merge(
        df_test_batch.alias("source"),
        "target.id = source.id"
    )
    .whenMatchedUpdateAll()
    .whenNotMatchedInsertAll()
    .execute()
)
```
